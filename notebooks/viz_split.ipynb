```python
def mlog_gauss_2d(mean, covar):
    H = np.linalg.inv(covar)
    #print(covar, H)
    sign, logdet = np.linalg.slogdet(H)
    logf = 0.5 * logdet - np.log(2. * np.pi)
    def f(x):
        arg = np.dot((x - mean), H)
        arg = np.dot(arg, (x - mean).T)
        #return np.exp(-0.5 * arg)
        #return -0.5 * np.diag(arg)
        return np.diag(arg), logf
    return f
```
